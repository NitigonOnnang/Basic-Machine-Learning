ขั้นตอนที่ 3: ดาวน์โหลดโมเดล YOLOv8 ที่ผ่านการเทรนมาแล้ว
YOLO มีโมเดลที่ผ่านการเทรน (Pre-trained Model) มาแล้วบนชุดข้อมูลขนาดใหญ่อย่าง COCO dataset ทำให้โมเดลสามารถตรวจจับวัตถุทั่วไปได้กว่า 80 ชนิด เช่น คน, รถยนต์, สุนัข, แมว เป็นต้น เราจะใช้โมเดล yolov8n.pt ซึ่งเป็นเวอร์ชันที่เล็กและรวดเร็ว เหมาะสำหรับเริ่มต้น

```python
from ultralytics import YOLO
model = YOLO('yolov8n.pt')
```

ขั้นตอนที่ 4: ทำนายผล (Inference) จากรูปภาพ
ในขั้นตอนนี้ เราจะใช้โมเดลที่โหลดมาเพื่อทำนายวัตถุในรูปภาพ สามารถใช้รูปภาพจาก URL หรืออัปโหลดไฟล์เองก็ได้ครับ

```python
image_url = 'https://ultralytics.com/images/bus.jpg'
results = model(image_url)
```

ขั้นตอนที่ 5: แสดงผลลัพธ์การทำนาย

```python
# แสดงผลลัพธ์การทำนาย
# โดย results[0].plot() จะวาด bounding box และ label ลงบนภาพต้นฉบับ
# ผลลัพธ์ที่ได้จะเป็นภาพในรูปแบบ numpy array
annotated_image = results[0].plot()

# แสดงภาพผลลัพธ์ใน Google Colab
from google.colab.patches import cv2_imshow
import cv2

# แปลงภาพ BGR (ที่ได้จาก plot) เป็น RGB เพื่อแสดงผลให้สีถูกต้อง
annotated_image_rgb = cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB)

# แสดงภาพ
cv2_imshow(annotated_image)
```

```python
# Import libraries
from ultralytics import YOLO
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow
import os
from IPython.display import Video, display
import numpy as np
```

ขั้นตอนที่ 2: ดาวน์โหลดวิดีโอตัวอย่าง
เพื่อให้ทุกคนสามารถรันโค้ดได้เหมือนกัน เราจะดาวน์โหลดวิดีโอตัวอย่างสั้นๆ มาไว้ใน Colab ก่อน

```python
import gdown

# ใช้ link แชร์จาก Google Drive
# ตัวอย่าง: https://drive.google.com/file/d/1ABC123XYZ/view?usp=sharing
file_id = '1Qu2SmP-8B6o6sRvqTp92lpbXiytNMOmY'  # เอา ID จาก URL

# ดาวน์โหลด
output = 'test.mp4'
gdown.download(f'https://drive.google.com/uc?id={file_id}', output, quiet=False)

video_path = output
```

ขั้นตอนที่ 3: โหลดโมเดล YOLOv8
เรายังคงใช้โมเดล yolov8n.pt ซึ่งเป็นโมเดลขนาดเล็กที่ผ่านการเทรนมาแล้ว เหมาะสำหรับงานที่ไม่ซับซ้อนและต้องการความรวดเร็ว

```python
# ====================================
# ส่วนที่ 2: โหลดโมเดล YOLO
# ====================================
# ใช้ YOLOv8n (nano) - เร็วที่สุด
# ตัวเลือกอื่นๆ: yolov8s.pt, yolov8m.pt, yolov8l.pt, yolov8x.pt (ใหญ่และแม่นยำกว่า)
model = YOLO('yolov8n.pt')
print("โหลดโมเดล YOLOv8 สำเร็จ!")
```

ขั้นตอนที่ 4: สั่งให้โมเดลทำนายวัตถุในวิดีโอ
นี่คือขั้นตอนที่สำคัญที่สุด เราจะส่งไฟล์วิดีโอเข้าไปให้โมเดลประมวลผล โดยโมเดลจะวิเคราะห์วิดีโอทีละเฟรม (Frame-by-frame) แล้วสร้างไฟล์วิดีโอผลลัพธ์ออกมา

```python
#video_path = 'test2.mp4'
# ====================================
# ส่วนที่ 3: ประมวลผลวีดีโอด้วย YOLO
# ====================================
print("\n🎬 กำลังประมวลผลวีดีโอ...")
print("⏳ กรุณารอสักครู่...")

# ใช้ YOLO predict ซึ่งจะบันทึกวีดีโอให้อัตโนมัติ
results = model.predict(
    source=video_path,
    save=True,              # บันทึกวีดีโอผลลัพธ์
    stream=True,
    conf=0.5,              # ความมั่นใจขั้นต่ำ (0-1)
    iou=0.5,               # IoU threshold
    show_labels=True,       # แสดงชื่อวัตถุ
    show_conf=True,        # แสดงความมั่นใจ
    line_width=2,          # ความหนาของเส้นกรอบ
    verbose=True           # แสดงความคืบหน้า
)
#results = model.predict(source=video, save=True, stream=True)
results = list(results)
print("✅ ประมวลผลเสร็จสิ้น!")
```

ขั้นตอนที่ 5: แสดงผลลัพธ์การทำนาย
ผลลัพธ์ที่ได้จากโมเดลจะอยู่ในตัวแปร results ซึ่งเราสามารถนำมาแสดงผลเป็นรูปภาพที่มีกรอบ (Bounding Box) และชื่อวัตถุ (Class Label) กำกับได้

```python
# ====================================
# ส่วนที่ 5: แสดงผลลัพธ์
# ====================================
# ดาวน์โหลดวีดีโอผลลัพธ์
output_path = results[0].save_dir
# แปลง .avi เป็น .mp4 ด้วย FFmpeg
input_file = output_path
output_file = 'detected_video.mp4'

!ffmpeg -i {input_file} -vcodec libx264 -acodec aac {output_file} -y

# แสดงวีดีโอ
from IPython.display import Video, display
display(Video(output_file, width=800, embed=True))
```
